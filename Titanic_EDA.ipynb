- **Missing Values**: Age (177), Cabin (687), Embarked (2)
- **Survival is higher** among:
  - Passengers in **1st class**
  - **Females**
  - Passengers with lower number of siblings/parents
- **Age** and **Fare** are slightly skewed with outliers.
- Strong correlation observed between:
  - Pclass and Fare (negatively)
  - Parch and SibSp (moderate)
